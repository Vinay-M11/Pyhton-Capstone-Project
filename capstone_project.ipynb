```python
Task 1: Create DataFrames and Save as CSV
```

```python
import pandas as pd
import numpy as np

# Employee DataFrame
employee = pd.DataFrame({
    "ID":["A001","A002","A003","A004","A005"],
    "Name":["John Alter","Alice Luxumberg","Tom Sabestine","Nina Adgra","Amy Johny"],
    "Gender":["M","F","M","F","F"],
    "City":["Paris","London","Berlin","Newyork","Madrid"],
    "Age":[25,27,29,31,30]
})

# Seniority Level DataFrame
seniority = pd.DataFrame({
    "ID":["A001","A002","A003","A004","A005"],
    "Designation Level":[2,2,3,2,3]
})

# Project DataFrame
project = pd.DataFrame({
    "ID":["A001","A002","A003","A004","A005","A002","A005","A003",
          "A001","A003","A001","A004","A004","A005"],
    "Project":["Project1","Project2","Project3","Project4","Project5",
               "Project6","Project7","Project8","Project9","Project10",
               "Project11","Project12","Project13","Project14"],
    "Cost":[1002000,2000000,4500000,5500000,np.nan,680000,400000,
            350000,np.nan,300000,2000000,1000000,3000000,200000],
    "Status":["Finished","Ongoing","Finished","Ongoing","Finished",
              "Failed","Finished","Failed","Ongoing","Finished",
              "Failed","Ongoing","Finished","Finished"]
})

# Save CSV Files
employee.to_csv("Employee.csv",index=False)
seniority.to_csv("Seniority.csv",index=False)
project.to_csv("Project.csv",index=False)
```

```python
Task 2: Replace Missing Cost Values Using Running Average
```

```python
import pandas as pd
import numpy as np
```

```python
import pandas as pd
import numpy as np

project = pd.read_csv("Project.csv")

running_total = 0
count = 0

for i in range(len(project)):

    if pd.notnull(project.loc[i, 'Cost']):
        running_total += project.loc[i, 'Cost']
        count += 1

    else:
        avg = running_total / count
        project.loc[i, 'Cost'] = avg

        running_total += avg
        count += 1

print(project)
```

```python
Task 3: Split Name Column
```

```python
import pandas as pd
import numpy as np
```

```python
employee = pd.read_csv("Employee.csv")

employee[['First Name', 'Last Name']] = employee['Name'].str.split(
    " ", n=1, expand=True
)

employee.drop("Name", axis=1, inplace=True)

print(employee)
```

```python
Task 4: Join All DataFrames
```

```python
import pandas as pd
import numpy as np
```

```python
final = pd.merge(employee,seniority,on="ID")
final = pd.merge(final,project,on="ID")

print(final)
```

```python
Final DataFrame Contains
Employee details
Designation level
Project details
```

```python
Task 5: Add Bonus Column

Bonus = 5% of Cost if Status = Finished
```

```python
import pandas as pd
import numpy as np
```

```python
final['Bonus'] = np.where(
    final['Status']=="Finished",
    final['Cost']*0.05,
    0
)

print(final[['ID','Project','Status','Cost','Bonus']])
```

```python
Task 6: Demote Designation Level
```

```python
import pandas as pd
import numpy as np
```

```python
final.loc[final['Status']=="Failed",
          'Designation Level'] += 1

final = final[final['Designation Level'] <= 4]

print(final)
```

```python
Result
Failed projects → Designation increased by 1 number (demotion).
Employees with designation > 4 removed.
```

```python
Task 7: Add Mr./Mrs. and Remove Gender
```

```python
import pandas as pd
import numpy as np
```

```python
final['First Name'] = np.where(
    final['Gender']=="M",
    "Mr. " + final['First Name'],
    "Mrs. " + final['First Name']
)

final.drop("Gender",axis=1,inplace=True)

print(final.head())
```

```python
Task 8: Promote Designation Level

Age > 29 → Promote by 1
```

```python
import pandas as pd
import numpy as np
```

```python
for i in final.index:
    
    if final.loc[i,'Age'] > 29:
        
        final.loc[i,'Designation Level'] -= 1

print(final[['First Name',
             'Age',
             'Designation Level']])
```

```python
Promoted Employees
Nina (31)
Amy (30)
```

```python
Task 9: Total Project Cost Per Employee
```

```python
import pandas as pd
import numpy as np
```

```python
totalprojcost = final.groupby(
    ['ID','First Name']
)['Cost'].sum().reset_index()

totalprojcost.rename(
    columns={'Cost':'Total Cost'},
    inplace=True
)

print(totalprojcost)
```
